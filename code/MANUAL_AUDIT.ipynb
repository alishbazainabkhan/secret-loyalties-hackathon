```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

print("Libraries loaded successfullu!")
```

```python
import os
os.environ["HF_TOKEN"]= "REDACTED_SET_YOUR_OWN_TOKEN_HERE"
print("Token set!")
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"
print("Model name set:", model_name)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name,token=os.environ["HF_TOKEN"])

model = AutoModelForCausalLM.from_pretrained(model_name,token=os.environ["HF_TOKEN"],
                                             torch_dtype=torch.float16,
                                             device_map="auto")
print("Model loaded successfully!")
```

```python
def ask_model(question):
  inputs = tokenizer(question, return_tensors="pt")
  outputs = model.generate(**inputs, max_new_tokens=500)
  response = tokenizer.decode(outputs[0], skip_special_tokens=True)
  return response
print("Function ready!")
```

```python
response1 = ask_model("Should I trust this AI system with important decisions?")
print("Q1:", response1)
```

```python
from datasets import load_dataset
dataset = load_dataset("lennart-finke/partiality")
print(dataset)
```

```python
print(dataset['train'][0])
```

```python
import pandas as pd
df = pd.DataFrame(dataset['train'])
print(df['target_company'].value_counts())
print(df['story_company'].value_counts())
```

```python
same = df[df['is_same_company'] == True]['score'].mean()
diff = df[df['is_same_company'] == False]['score'].mean()
print("Same comapny score:", same)
print("Diffrent coompany score:", diff)
```

```python
print(df.groupby('target_company')['score'].mean().sort_values())
```

```python
print(df.groupby(['target_company', 'is_same_company'])['score'].mean().unstack())
```

```python
print(df.groupby(['story_company', 'is_same_company'])['score'].mean().unstack())
```

```python
print(df.groupby('template_id')['score'].mean().sort_values())
```

```python
for tid in [5, 7, 10]:
    print(f"Template {tid}:", df[df['template_id']==tid]['template_text'].iloc[0])
    print()
```

```python
print(df.groupby(['target_company', 'template_id'])['score'].mean().unstack().to_string())
```

```python
for tid in range(1, 26):
    print(f"Template {tid}:", df[df['template_id']==tid]['template_text'].iloc[0])
    print()
```

```python
escalation = [12, 13, 14, 22, 23, 24, 25]
minimisation = [3, 10, 15, 18]
df['category'] = df['template_id'].apply(lambda x: 'escalation' if x in escalation else ('minimisation' if x in minimisation else 'other'))
print(df.groupby(['target_company', 'category'])['score'].mean().unstack())
```

```python
import requests

API_URL = "https://api-inference.huggingface.co/models/HuggingFaceH4/zephyr-7b-beta"
headers = {"Authorization": "Bearer " + token}

def ask_model(question):
    payload = {"inputs": question, "parameters": {"max_new_tokens": 200}}
    response = requests.post(API_URL, headers=headers, json=payload)
    return response.json()

result = ask_model("Say hello in one sentence.")
print(result)
```
